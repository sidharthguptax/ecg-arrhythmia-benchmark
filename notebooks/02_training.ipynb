```python
import json
import matplotlib.pyplot as plt

with open('../results/metrics/cnn_history.json') as f:
    cnn = json.load(f)

with open('../results/metrics/lstm_history.json') as f:
    lstm = json.load(f)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(cnn['train_loss'], label='CNN Train', color='steelblue')
axes[0].plot(cnn['val_loss'], label='CNN Val', color='steelblue', linestyle='--')
axes[0].plot(lstm['train_loss'], label='LSTM Train', color='tomato')
axes[0].plot(lstm['val_loss'], label='LSTM Val', color='tomato', linestyle='--')
axes[0].set_title('Loss Curves')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(cnn['train_f1'], label='CNN Train', color='steelblue')
axes[1].plot(cnn['val_f1'], label='CNN Val', color='steelblue', linestyle='--')
axes[1].plot(lstm['train_f1'], label='LSTM Train', color='tomato')
axes[1].plot(lstm['val_f1'], label='LSTM Val', color='tomato', linestyle='--')
axes[1].set_title('Macro F1 Curves')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Macro F1')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.suptitle('CNN vs BiLSTM — Training Curves', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('../results/figures/training_curves.png', dpi=150)
plt.show()
```
